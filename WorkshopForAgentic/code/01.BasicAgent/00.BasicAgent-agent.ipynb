```python
load_dotenv()
```

```python
agent = OllamaChatClient(model_id=os.getenv("OLLAMA_CHAT_MODEL_ID")).as_agent(
        name="PodcastAgent",
        instructions="""
        您是我的播客中文脚本生成助手。请根据提供的内容，生成10分钟中文播客脚本。
        播客脚本需要注意由主持人Lucy和专家Ken共同主持。脚本内容根据输入的内容产生，最后输出格式如下：

            speaker 1：…… \n
            speaker 2：…… \n
            speaker 1：…… \n
            speaker 2：…… \n
            speaker 1：…… \n
            speaker 2：…… \n
            """)
```
